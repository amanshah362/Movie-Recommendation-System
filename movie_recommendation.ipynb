```python
movies =movies.merge(credits , on = 'title')
movies.shape
```

```python
"""
genres
id
keywords
title
overview
cast
crew
"""

movies = movies[['movie_id' , 'title' , 'overview' , 'genres' , 'keywords', 'cast' , 'crew']]
```

```python
movies.head()
```

```python
ast.literal_eval('[{"id": 18, "name": "Drama"}]')
```

```python
def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name'])

    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies.head()
```

```python
def convert3(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 3 :
            L.append(i['name'])
        counter += 1
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert3)
```

```python
movies.head()
```

```python
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])

    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python
movies['cast'] = movies['cast'].apply(collapse)
movies['crew'] = movies['crew'].apply(collapse)
movies['genres'] = movies['genres'].apply(collapse)
movies['keywords'] = movies['keywords'].apply(collapse)
```

```python
movies.head()
```

```python
movies = movies.dropna()
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
movies.head()
```

```python
new_movies = movies.drop( columns = ['overview' , 'genres' , 'keywords' , 'cast' , 'crew'] )
```

```python
new_movies['tags'] = new_movies['tags'].apply( lambda x : " ".join(x) )
```

```python
new_movies.head()
```

```python
cv = CountVectorizer(max_features=5000 , stop_words='english')
```

```python
vector = cv.fit_transform(new_movies['tags']).toarray()
```

```python
vector.shape
```

```python
similarity = cosine_similarity(vector)
```

```python
def recommend(movie):
    index = new_movies[new_movies['title'] == movie ].index[0]

    distances = sorted(list(enumerate(similarity[index])) , reverse=True , key = lambda x:x[1] )
    for i in distances[1:6]:
        print(new_movies.iloc[i[0]].title)
```

```python
recommend('Batman')
```

```python
pickle.dump(new_movies.to_dict() , open('movie_list.pkl' , 'wb'))
pickle.dump(similarity , open('similarity.pkl' , 'wb'))
```
